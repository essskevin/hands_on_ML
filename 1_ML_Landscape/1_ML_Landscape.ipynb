## 1. Introduction

Thanks to big data, hardware improvements and algorithmic innovations, with time ML went from powering services in the background to become the service itself. 

But, what is ML?
We have bunches of ways for defining ML:
- ML is the science of programming computers so they can learn from data;
- ML is the field of study giving computers ability to learn without being explicitly programmed;
- a computer program is said to learn from experience E with respect to some task T and some performance measure P, if its performance on T, as measured by P, improves with experience E.

This last definition helps us define some ML terms.
Suppose we take as an example of ML program a Spam Filter: in this case the task T is to flag spam for new emails; the examples the system uses to learn are called **Training Set**, which define the experience E; each of these example is called a **Training Instance** (or **Sample**); the part that learns and make predictions is called **Model**; we can then measure the performance P of this model, for example by using the **accuracy metric** (ratio of correctly classified emails).

But, why do we use ML?
- Complex problems with long list of rules (e.g., a spam filter program using traditional programming techniques would take to write a long list of complex rules, which are pretty hard to maintain, while ML allows the system to automatically learns those rules, taking to an easier to maintain program)
- Fluctuating environments to keep up to date;
- Getting insights about complex problems and large amounts of data (ML models can be inspected helping human to learn from them).

## 3. Types of ML Systems
We can classify ML systems based on different criteria:
- **Training Supervision** (i.e., how they are guided during training);
- **Online vs Batch Learning** (i.e., whether or not they can learn incrementally on the fly);
- **Instance-Based vs Model-Based** (i.e., whether they simply compare new points to known or instead they detect patterns).

Let's deep dive in all of these:


### Training Supervision

It reguards the amount/type of supervision the system get during training. The main categories are:
- **Supervised Learning**: the algorithm is fed with a training set containing the desired solutions, called **Labels**. The 2 main type of Supervised Learning tasks are **Classification**, which involves to predict a discrete target value (e.g., a spam filter asked to predict "Spam" or "Not Spam"), and **Regression**, which involves to predict a numeric target value (e.g., a price prediction system for cars); in both cases, the training set contains labels to allow the system to learn from them;
- **Unsupervised Learning**: the training set is unlabeled. The main type of Unsupervised Learning tasks are **Clustering**, which involves to detect groups of similar instances, **Visualization**, which involves to get a representation of data, **Dimensionality Reduction**, which is strictly related to the last one since it involves to simplify data without losing too much information and **Anomaly Detection**, which involves to detect anomalies given mostly normal instances. In all of these cases, nobody tells the system which the solution (label) is;
- **Semi-Supervised Learning**: the training set is partially labeled and partially unlabeled. This is an usual task when getting labels is time-consuming and costly. Most of these are combinations of Unsupervised and Supervised Algorithms (e.g., Google Photos has a Clustering algorithms to segment people in your photos; if you then tell who these people are, it uses these labels for every similar "object");
- **Self-Supervised Learning**: getting a labeled dataset from an unlabeled one (once is labeled, any Supervised Learning algorithm can be used). Usually, the resulting model is not the final goal, but just an initial model you can tweak (this is known as **Fine-Tuning**) to tackle a slightly different task (e.g., we want a pet classification model, but we don't have labels: we train an image-repairing model; once it performs well, it should be able to recognize the pet species, so we fine.tune on a smaller labeled dataset). As we will see, **Large Language Models (LLMs)** are trained in a similar way by masking random words and training a model to predict the missing word; then it is fine-tuned for plenty of applications.
- **Reinforcement Learning**: the learning system observe the environment and perform actions, getting **rewards** (or **penalties**) in return. This takes to learn the best strategy, called a **policy**.

### Batch vs Online Learning
It reguards whether the system can or not learn incrementally from a stream of incoming data:
- **Batch** (or **Offline**) **Learning**: system must be trained on the full **Batch** of data (this is generally resource demanding, so it is typically done **Offline**). Once it has been trained, it can be launched into production and it won't learn anymore. This generally takes the model's performance to decay over time due to the **Data Drift** phenomenon (world evolves while model not). A solution to this is to regularly re-train the model on up-to-date data. Of course, this last solution can be a problem when we need quick updates. An example of this type of models is **Random Forest**;
- **Online Learning**: system can be incrementally trained by feeding data sequentially both individually or in small groups (called **Mini-Batches**). This is a good option when we need a rapid system, or when the computing resources are limited (e.g., when the dataset won't fit in one machine's memory). An important parameter of Online Learning is the **Learning Rate** (i.e., how fast the system should adapt to changing data): the higher it is, the more rapid the system will be (meaning it will also tend to quickly forget old data: **Catastrophic Forgetting**). An example of this type of models is **Gradient Descent**.

### Instance-Based vs Model-Based Learning
It reguards whether the system simply compares new points to known or detects patterns (i.e., how it generalizes):
- **Instance-Based Learning**: system learns the examples and the generalizes new cases using similarity measures to compare new cases to learned ones. This type of systems shines with small datasets, but does not scale well to both number of instances/features;
- **Model-Based Learning**: system builds a model from the examples and then use it to make predictions.

## 4. Main Challenges of ML
Since ML is mostly about selecting a model and train it on some data, the 2 things that can go wrong are "bad model" and "bad data".

### Bad Data Examples

#### Insufficient Quantity of Training Data
While a child requires one (or a bunch) of apples to understand what an apple is, a ML system is not like that: it takes a lot of data for most ML algorithms to work properly, even in simpler tasks.

Extreme examples of the usefulness of data are the papers:
- [Scaling to very large corpora for natural language disambiguation](https://dl.acm.org/doi/10.3115/1073012.1073017): this paper by Microsoft researchers shows how training very different ML algorithms, including fairly simples ones, performed almost identically on complex problem of Natural Language Disambiguation once they were given enough data;
- [The Unreasonable Effectiveness of Data](https://static.googleusercontent.com/media/research.google.com/it//pubs/archive/35179.pdf): this paper popularized the idea that data matters more than algorithms for complex problems.

It should still be noticed how small and medium-sized datasets are very common and how it is not always easy or cheap to get extra data.

#### Nonrepresentative Training Data
It is crucial to use a training set that is representative of the cases you want to generalize to. Not doing so is known as **Sampling Bias**; one of the most famous cases of this is the 1936 US presidential election Landon vs Roosevelt, in which a very large poll was conducted: 10 million mails had been sent and 2.4 million got an answer, predicting with high confidence that Landon would get 57%. At the end, Roosevelt won with 62%. This is due to the sampling method: first, the addresses to which to send mails was obtained througb telephone directories, magazine subscribers, club memberships and other things mainly used by wealthier people, more likely to vote Republican (Landon); second, only 25% of these answered, potentially ruling out people that didn't care much about politics (this is known as **Nonresponse Bias**).

#### Poor-Quality Data
If training data is full of errors, outliers, noise, it would be harder for the system to detect patterns. That's why Data Scientists spend a significant part of their time cleaning data, for example by discarding outliers/missing values, or trying to fixing them (scaling, transforming, replacing).

#### Irrelevant Features
"Garbage in, Garbage Out". A critical part of a ML project is coming up with a good set of features to train on; this process, known as **Feature Engineering** involves the following steps:
- **Feature Selection**: select most useful features;
- **Feature Extraction**: combine features to produce a more useful one (e.g., dimensionality reduction);
- Creating new features with new data.

### Bad Model Examples

#### Overfitting the Training Data
**Overfitting** is when a model performs well on the training data, but does not generalize well (think of you being robbed by a taxi driver in a foreign country and overgeneralizing saying that all taxi drivers in that country are thieves). The more complex a model is, the more likely it will also learn by the noise of the data (this learned pattern is due to chance and won't generalize to new instances), taking to overfitting. Possible solutions to overfitting are:
- Simplifying the model by taking one with fewer parameters (e.g., lower-degree polynomial model);
- Getting more training data;
- Reduce noise in data (e.g., fixing data errors and remove outliers).

Constraining a model to make it simpler to fight overfitting is called **Regularization**.

#### Underfitting the Training Data
**Underfitting** is when a model is too simple to learn the underlying structure of data, taking it to perform bad even on the training set. Possible solutions to underfitting are:
- Select a more powerful model;
- Feature Engineering;
- Reduce Regularization.

### Other Problems
Even if you have a large and clean dataset and a beautiful model, there are other problems that you could encounter. For example, you could encounter some **Deployment Issues**, such as model too complex to maintain, too large to fit in memory, too slow, not able to scale, with security vulnerabilities, outdated, and so on. These tasks are usually far from Data Modeling, so companies usually have a dedicated **MLOps (ML Operations)** team.

## Testing and Evaluating
The only way to know if a model will generalize well on new cases is to actually try it out on new cases.

One option for doing so is to deploy the model and monitor it; this works, but if the model is bad, users will complain, so it is not the best idea.

Another better option is to split your data into 2 sets (this is called **Hold-Out**):
- **Training Set**, on which you train the model;
- **Test Set**, on which you test the model.

The error you get by training the model on just the training set and testing it on the test set is an estimate of the **Generalization Error** (i.e., the error you would get on new cases). Usually, 80% of data is used for training, 20% for testing (however, when dataset size increases, even keeping 1% for validation is enough to get good estimates).

Now, suppose to train models aiming to minimize error on this Test Set, adapting the model to do so. The problem of this approach is that you're adapting the model to that specific Test Set, so your model won't perform as well on new data.

A solution for this is to furtherly split the train set into 2 sets (this is called **Hold-Out Validation**):
- **Training Set**, on which you train the model;
- **Validation Set**, on which you evaluate the model.

You can now focus on finding the model that trained on this Training Set performs best on the Validation Set. Once you've found it, you train it on the Full Training Set (Training Set + Validation Set) and test it on the Test Set. This will be the estimate of the Generalization Error.

Another approach is **Cross-Validation**, in which we train the model on one portion of the data and evaluate it on the other portions; we then swap the roles, training the model on one portion of the data we used to evaluate it, and evaluate it on the other portions and so on. We then average out all the metrics, ending up with a more accurate measure of its Generalization Error. This is generally the best possible approach, but it is also time and resource demanding (we must train multiple models).
